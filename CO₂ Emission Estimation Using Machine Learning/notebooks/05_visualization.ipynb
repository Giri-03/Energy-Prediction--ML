```python
import matplotlib.pyplot as plt
import numpy as np

# The variables 'y_test' and 'y_pred_rf' are not defined in this cell.
# Please run the cells that generate model predictions to define them.
# For demonstration purposes, dummy data is created here.
np.random.seed(42)
y_test = np.linspace(100, 500, 100)  # Dummy actual values
y_pred_rf = y_test + np.random.normal(0, 25, 100)  # Dummy predicted values

plt.figure(figsize=(8,6))
plt.scatter(y_test, y_pred_rf, color='blue', alpha=0.6)
plt.xlabel("Actual CO₂ Emissions")
plt.ylabel("Predicted CO₂ Emissions")
plt.title("Actual vs Predicted CO₂ Emissions")
plt.show()
```
